```python
df = pd.read_csv('position_level_salary_5000_rows.csv')
x = df.iloc[:, 1:2].values
y = df.iloc[:, 2].values
```

```python
# Our_Data
df.head()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(x, y)
plt.show()
```

```python
# Fitting linear regression to dataset
from sklearn.linear_model import LinearRegression
lin_regs = LinearRegression()
lin_regs.fit(x, y)
```

```python
# Fitting Polynomial regression to dataset
from sklearn.preprocessing import PolynomialFeatures
poly_regs = PolynomialFeatures(degree = 3)
x_poly = poly_regs.fit_transform(x)
lin_regs_2 = LinearRegression()
lin_regs_2.fit(x_poly, y)
```

```python
# Visualizing result for Linear Regression Model
plt.scatter(x, y, color = 'red')
plt.plot(x, lin_regs.predict(x), color = 'blue')
plt.title('Truth or Bluff (Linear Regression)')
plt.xlabel('Position Level')
plt.ylabel('Salary')
plt.show()
```

```python
# Visualizing result for Polynomial Regression Model
plt.scatter(x, y, color = 'red')
plt.plot(x, lin_regs_2.predict(poly_regs.fit_transform(x)), color = 'blue')
```

```python
# Predicting a new result with Linear Regression
prediction = lin_regs.predict([[1]])
print(prediction)
```

```python
polyPprediction = lin_regs_2.predict(poly_regs.fit_transform([[1]]))
print(polyPprediction)
```
